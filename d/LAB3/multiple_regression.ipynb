# LAB 2: Multiple Linear Regression — F1 Drivers

**วิชา:** Machine Learning — ใบงานที่ 3 (Regression & Classification)

**เป้าหมาย:** ใช้ **หลาย Feature พร้อมกัน** ทำนายจำนวนชนะ (Career Wins) ของนักแข่ง F1

**ต่างจาก Simple Regression:**
- Simple = ใช้ 1 ตัวแปร (pole positions อย่างเดียว)
- Multiple = ใช้หลายตัวแปร (pole + podiums + points + championships)

ยิ่งใช้ข้อมูลหลายด้านประกอบกัน โมเดลยิ่งทำนายได้แม่นยำขึ้น

## 1. Import Libraries
เพิ่ม `r2_score` และ `mean_absolute_error` สำหรับวัดความแม่นยำของโมเดล

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error
```

## 2. Load Data (อ่านข้อมูล)

```python
df = pd.read_csv("f1_drivers.csv")
print("ขนาดข้อมูล (แถว, คอลัมน์):", df.shape)
df.head()
```

## 3. Prepare Data (เตรียมข้อมูล — จุดต่างสำคัญ!)

**x (Features)** = ใช้ **4 คอลัมน์พร้อมกัน**: pole_positions, career_podiums, career_points, championships

**y (Target)** = career_wins

> สังเกต: ใน `df[[...]]` เราใส่หลายคอลัมน์คั่นด้วยจุลภาค นี่คือหัวใจของ Multiple Regression

```python
# ใช้หลาย Feature พร้อมกัน
features = ["pole_positions", "career_podiums", "career_points", "championships"]
x = df[features]
y = df["career_wins"]

print("Features ที่ใช้:", features)
x.head()
```

## 4. Train Model (สร้างและสอนโมเดล)
โมเดลจะหาค่าน้ำหนัก (ความชัน) ของแต่ละ Feature แยกกัน

```python
model = LinearRegression()
model.fit(x, y)

# แสดงน้ำหนักของแต่ละ Feature
print("น้ำหนัก (ความชัน) ของแต่ละ Feature:")
for feature, coef in zip(features, model.coef_):
    print(f"  {feature}: {round(coef, 4)}")
print("\nจุดตัดแกน (c):", round(model.intercept_, 4))
```

## 5. Predict (ทำนาย)
ทดลองทำนายนักแข่งสมมติที่มี: Pole 30, Podium 60, Points 1500, แชมป์ 2 สมัย

```python
# สร้างข้อมูลใหม่ (มีชื่อคอลัมน์ตรงกับตอน train เพื่อกัน warning)
new_driver = pd.DataFrame({
    "pole_positions": [30],
    "career_podiums": [60],
    "career_points": [1500],
    "championships": [2]
})

pred = model.predict(new_driver)
print("นักแข่งคนนี้คาดว่าชนะ:", round(pred[0], 1), "ครั้ง")
```

## 6. Evaluate Model (วัดความแม่นยำ)

- **R² Score:** เข้าใกล้ 1 = โมเดลดี (อธิบายข้อมูลได้มาก)
- **MAE:** ยิ่งน้อยยิ่งดี (ทำนายพลาดเฉลี่ยกี่ครั้ง)

```python
y_pred = model.predict(x)

r2 = r2_score(y, y_pred)
mae = mean_absolute_error(y, y_pred)

print("R2 Score:", round(r2, 4), "(เข้าใกล้ 1 = ดี)")
print("MAE:", round(mae, 4), "(ยิ่งน้อยยิ่งดี)")
```

## 7. Visualize (เปรียบเทียบค่าจริง vs ค่าทำนาย)

- แกน X = ค่าจริง (Actual Wins)
- แกน Y = ค่าที่โมเดลทำนาย (Predicted Wins)
- เส้นประ = เส้นทำนายสมบูรณ์แบบ (ถ้าจุดเกาะเส้นนี้ = ทำนายแม่น)

```python
plt.scatter(y, y_pred, color='blue', label='Predictions')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', label='Perfect Prediction')
plt.xlabel('Actual Wins')
plt.ylabel('Predicted Wins')
plt.title('Multiple Regression: Actual vs Predicted')
plt.legend()
plt.show()
```

## สรุป

- Multiple Regression ใช้ **หลาย Feature** ทำนายพร้อมกัน → แม่นกว่า Simple
- ค่าน้ำหนัก (coef) แต่ละตัวบอกว่า Feature นั้นมีผลต่อผลลัพธ์มากน้อยแค่ไหน
- วัดคุณภาพด้วย **R² Score** (ยิ่งใกล้ 1 ยิ่งดี) และ **MAE** (ยิ่งน้อยยิ่งดี)
- ขั้นถัดไป: ลอง **Classification** — ทำนายว่านักแข่งคนนี้เป็น "แชมป์โลกหรือไม่"
