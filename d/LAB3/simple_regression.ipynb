# LAB 1: Simple Linear Regression — F1 Drivers

**วิชา:** Machine Learning — ใบงานที่ 3 (Regression & Classification)

**เป้าหมาย:** ใช้ Simple Linear Regression ทำนายจำนวนชนะ (Career Wins) จากจำนวน Pole Positions ของนักแข่ง F1

**แนวคิด:** นักแข่งที่ออกตัวในตำแหน่งแรก (Pole) บ่อย มักมีโอกาสชนะสูงตามไปด้วย เราจะหาเส้นตรงที่อธิบายความสัมพันธ์นี้

## 1. Import Libraries
เรียกใช้เครื่องมือที่จำเป็น: `pandas` (จัดการข้อมูล), `matplotlib` (วาดกราฟ), `scikit-learn` (โมเดล ML)

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

## 2. Load Data (อ่านข้อมูล)
อ่านไฟล์ `f1_drivers.csv` เข้ามาเป็น DataFrame แล้วดูขนาดข้อมูลและตัวอย่าง 5 แถวแรก

```python
df = pd.read_csv("f1_drivers.csv")
print("ขนาดข้อมูล (แถว, คอลัมน์):", df.shape)
df.head()
```

## 3. Prepare Data (เตรียมข้อมูล)
- **x (Feature)** = `pole_positions` — สิ่งที่เรารู้ ใช้เป็นตัวทำนาย
- **y (Target)** = `career_wins` — สิ่งที่อยากทำนาย

> หมายเหตุ: x ต้องเป็น 2 มิติ จึงใช้วงเล็บสองชั้น `df[[...]]`

```python
x = df[["pole_positions"]]   # Feature (2 มิติ)
y = df["career_wins"]        # Target
```

## 4. Train Model (สร้างและสอนโมเดล)
สร้างโมเดล `LinearRegression` แล้วสั่ง `.fit()` เพื่อให้โมเดลเรียนรู้ความสัมพันธ์ (หาค่าความชัน m และจุดตัดแกน c ที่ดีที่สุด)

```python
model = LinearRegression()
model.fit(x, y)

print("ความชัน (m):", model.coef_[0])
print("จุดตัดแกน (c):", model.intercept_)
```

## 5. Predict (ทำนาย)
ทดลองทำนาย: ถ้านักแข่งมี Pole Position 50 ครั้ง โมเดลคาดว่าจะชนะกี่ครั้ง?

```python
pred = model.predict([[50]])
print("Pole 50 ครั้ง คาดว่าชนะ:", round(pred[0], 1), "ครั้ง")
```

## 6. Visualize (วาดกราฟ)
- จุดสีน้ำเงิน = ข้อมูลจริง
- เส้นสีแดง = เส้นทำนายของโมเดล

ถ้าเส้นแดงพาดผ่านกลุ่มจุดได้ดี แปลว่าโมเดลอธิบายข้อมูลได้ดี

```python
plt.scatter(x, y, color='blue', label='Actual Data')
plt.plot(x, model.predict(x), color='red', label='Prediction Line')
plt.xlabel('Pole Positions')
plt.ylabel('Career Wins')
plt.title('F1: Pole Positions vs Career Wins')
plt.legend()
plt.show()
```

## สรุป
- โมเดลเรียนรู้เส้นตรง `wins = m × pole + c`
- ค่า **m (ความชัน)** บอกว่า Pole เพิ่ม 1 ครั้ง ชนะเพิ่มกี่ครั้งโดยเฉลี่ย
- ขั้นถัดไป: ลองทำ **Multiple Linear Regression** โดยใช้หลาย Feature พร้อมกัน (เช่น pole + podiums + points) เพื่อทำนายได้แม่นขึ้น
